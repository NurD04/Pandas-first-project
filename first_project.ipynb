```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df.dtypes
```

```python
kpi = {
    "total sales" : df["Sales"].sum(),
    "total discount" : df["Discount"].sum(),
    "total profit" : df["Profit"].sum(),
    "orders" : df["Order ID"].nunique(),
    "average sales" : df["Sales"].mean(),
    "average profit" : df["Profit"].mean(),
}

kpi
```

```python
print("Grouping by category")
display(df.groupby("Category")["Profit"].sum())
df.groupby("Sub-Category")["Profit"].sum().sort_values()
```

```python

print("Sales for month")
display(df.set_index("Order Date").resample("ME")["Sales"].sum())
print("Profits for month")
display(df.set_index("Order Date").resample("ME")["Profit"].sum().sort_values())
```
